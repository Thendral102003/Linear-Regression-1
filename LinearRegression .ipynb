```python
x=data[['Age','Sleep Hours','Physical Activity (hrs/week)','Caffeine Intake (mg/day)','Alcohol Consumption (drinks/week)','Stress Level (1-10)','Heart Rate (bpm)','Breathing Rate (breaths/min)','Sweating Level (1-5)','Therapy Sessions (per month)','Diet Quality (1-10)']]
y=data['Anxiety Level (1-10)']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
x_train
```

```python
y_train
```

```python
from sklearn.linear_model import LinearRegression
LR=LinearRegression()
LR.fit(x_train,y_train)
```

```python
y_pre=LR.predict(x_test)
```

```python
y_pre
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
print(mean_squared_error(y_test,y_pre))
```

```python
print(mean_absolute_error(y_test,y_pre))
```
